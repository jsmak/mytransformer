### Layer Normalization 설명 및 결과 분석

**Layer Normalization (LayerNorm)** 은 딥러닝 모델의 학습 안정성과 속도 향상을 위해 사용되는 정규화 기법입니다.   
각 레이어의 입력에 대해 **평균을 0으로, 분산을 1로** 조정합니다.

**코드 설명:**
`(batch_size, seq_len, d_model)` 형태의 랜덤 텐서 `x`에 `nn.LayerNorm`을 적용합니다.   
`d_model=512`이므로, 각 시퀀스(512차원 벡터)에 대해 정규화가 수행됩니다.

**결과 (`standard_output`):**
```
(1, 5, 512)
평    균: [[-0.  0.  0.  0.  0.]]
표준편차: [[1. 1. 1. 1. 1.]]
```
*   각 시퀀스(512차원 벡터)의 `d_model` 차원에 대한 평균은 0에 가깝고, 표준편차는 1에 가깝게 정규화되었음을 확인할 수 있습니다.

````python
import torch
import torch.nn as nn

# 하이퍼파라미터 설정
d_model = 512 # 입력 차원

# LayerNorm인스턴스
layer_norm = nn.LayerNorm(normalized_shape=d_model)

#랜덤 값 생성
batch_size = 1
seq_len = 5

x = torch.randn(batch_size, seq_len, d_model)
input = x.detach().cpu().numpy()
print("입력")
print("평    균:",input.mean(axis=-1).round(3))
print("표준편차:",input.std(axis=-1).round(3))

normalized_x = layer_norm(x)
normalized_x = normalized_x.detach().cpu().numpy()
print("Layer Norm 적용 후")
print(normalized_x.shape)  # [batch_size, seq_len, d_model]
print("평    균:",normalized_x.mean(axis=-1).round(3)) # 평    균: [[ 0. 0.  0.  0. 0.]]
print("표준편차:",normalized_x.std(axis=-1).round(3)) # 표준편차: [[1. 1. 1. 1. 1.]]
````
