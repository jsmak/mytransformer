### **`PositionwiseFeedForward` 클래스:**
-   두 개의 선형 변환(`w_1`, `w_2`)과 ReLU 활성화 함수이루어져있습니다.
    -   `d_model`: 입력 및 출력 특징의 차원.
    -   `d_ff`: 피드포워드 네트워크의 은닉층 차원 (일반적으로 `d_model`보다 큽고, 보통 4배로 설정합니다).

### **코드 설명:**
-   `batch_size`, `seq_len`, `d_model`, `d_ff`는 예시 입력 텐서의 차원을 설정합니다.
-   `(batch_size, seq_len, d_model)` 형태의 무작위 텐서 `x`가 입력으로 생성됩니다.
-   `ff` 모듈이 `x`를 처리하고, 결과 `output` 텐서의 크기가 출력됩니다. 출력 크기는 `(8, 50, 512)`여야 합니다.
- 입력과 출력의 크기는 동일합니다.

```python
import torch
import torch.nn as nn

class PositionwiseFeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.w_1 = nn.Linear(d_model, d_ff)
        self.w_2 = nn.Linear(d_ff, d_model)
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.w_2(self.relu(self.w_1(x)))

batch_size = 8
seq_len = 50
d_model = 512
d_ff = 2048

ff = PositionwiseFeedForward(d_model, d_ff)
x = torch.randn(batch_size, seq_len, d_model)
print(x.size())
output =ff(x)
print(output.size())
```
